```python
import matplotlib.pyplot as plt
plt.style.use("ggplot")
```

```python
x = [2,3,4,8]
y_tiny_base = [78.8] * len(x)
y_tiny_moe = [80.8,81.0,81.1,78.1]
y_small_base = [89.1] * len(x)
y_small_moe = [ 89.7,88.9,89.1,89.1]
```

```python
plt.figure()
plt.plot(x, y_tiny_moe, '-o', color='tab:red')
plt.plot(x, y_small_moe, '-o', color='tab:blue')
plt.plot(x, y_tiny_base, '--', color='tab:red')
plt.plot(x, y_small_base, '--', color='tab:blue')
plt.legend(["MoE Tiny", "MoE Small", "Standard Tiny", "Standard Small"])
plt.xlabel("Liczba ekspertów")
plt.ylabel("Dokładność TOP 1")
plt.show()
```
